```python
#TASK_A
from collections import deque
N,M=map(int,input().split())
course={x:[] for x in range(1,N+1)}
indeg=[0]*N
for i in range(M):
    A,B=map(int,input().split())
    course[A].append(B)
    indeg[B-1]+=1

Q=deque()
for i in range (N):
    if indeg[i]==0:
        Q.append(i+1)

res=[]
while Q:
    node=Q.popleft()
    res.append(node)
    for i in course[node]:
        indeg[i-1]-=1
        if indeg[i-1]==0:
            Q.append(i)

if len(res)!=N:
    print(-1)
else:
    print(*res)
```

```python
#Task_B
from collections import deque
def bipartite(edge,source,visited):
    count=[0,0]
    Q=deque()
    visited[source]=0
    count[0]+=1
    Q.append(source)
    while Q:
        node=Q.popleft()
        for i in edge[node]:
            if visited[i]==-1:
                visited[i]= 1-visited[node]
                count[visited[i]]+=1
                Q.append(i)
            elif visited[i]==visited[node]:
                return 0
    return max(count)
N,M = map(int,input().split())
edge= {x:[] for x in range(1,N+1)}
visited= {x:-1 for x in range(1,N+1)}
p=0
for i in range(1,M+1):
  node1,node2 = map(int,input().split())
  edge[node1].append(node2)
  edge[node2].append(node1)
for i in range(1, N+1):
  if visited[i]== -1:
    p+=bipartite(edge, i,visited)
print(p)
```

```python
#Task_C
from collections import deque
def knight_moves(N,x1,y1,x2,y2):
  Q=deque()
  visited[x1-1][y1-1]= 1
  allowed = []
  for dx in [-2,-1,1,2]:
    for dy in [-2,-1,1,2]:
        if abs(dx)!=abs(dy):
            allowed.append((dx, dy))
  Q.append((x1,y1,0))
  while Q:
    x,y,possible_move= Q.popleft()
    if x==x2 and y==y2:
      return possible_move
    for X,Y in allowed:
      m1,m2 =x+X, y+Y
      if (1<=m1<=N and 1<=m2<=N) and visited[m1-1][m2-1]== 0:
        visited[m1-1][m2-1]=1
        Q.append((m1,m2,possible_move+1))
  return -1
N=int(input())
x1,y1,x2,y2 = map(int,input().split())
visited=[[0]*N for x in range(N)]
print(knight_moves(N,x1,y1,x2,y2))
```

```python
#task_D
from collections import deque
N=int(input())
edge= {x:[] for x in range(1,N+1)}
for x in range(N-1):
  node1,node2 =map(int,input().split())
  edge[node1].append(node2)
  edge[node2].append(node1)
def BFS(edge,source,N):
  visited= {x:0 for x in range(1,N+1)}
  leng=[0]*N
  Q= deque()
  visited[source]= 1
  Q.append(source)
  longest=source
  while Q:
    node=Q.popleft()
    for i in edge[node]:
      if visited[i]== 0:
        visited[i]= 1
        leng[i-1]=leng[node-1]+1
        Q.append(i)
        if leng[i-1]>leng[longest-1]:
          longest=i
  return longest,leng[longest-1]
A,d= BFS(edge,1,N)
B,d= BFS(edge,A,N)
print(d)
print(A,B)
```

```python
#Task_E
from collections import defaultdict

N=int(input())
words=[input().strip() for i in range(N)]

graph=defaultdict(list)
indeg={chr(i):0 for i in range(ord('a'), ord('z')+1)}
flag= True

for i in range(N-1):
    w1,w2=words[i],words[i+1]
    if w1.find(w2)==0 and len(w1)>len(w2):
        flag= False
        break
    for c1,c2 in zip(w1,w2):
        if c1!=c2:
            if c2 not in graph[c1]:
                graph[c1].append(c2)
                indeg[c2]+=1
            break

if not flag:
    print(-1)
else:
    w=[c for c in indeg if indeg[c]==0]
    w.sort()
    res= []
    used= {c for x in words for c in x}
    while w:
        u= w.pop(0)
        if u in used:
            res.append(u)
        for v in sorted(graph[u]):
            indeg[v]-= 1
            if indeg[v]== 0:
                w.append(v)
        w.sort()

    if len(res)<len(used):
        print(-1)
    else:
        for c in res:
          print(c, end='')
        print()
```

```python
#Task_F
from collections import deque
N,M,S,Q=map(int,input().split())
graph=[[] for i in range(N+1)]
for i in range(M):
    u,v=map(int,input().split())
    graph[u].append(v)
    graph[v].append(u)
sources=list(map(int,input().split()))
dests=list(map(int,input().split()))
dist=[-1]*(N+1)
Q= deque()
for i in sources:
    dist[i]= 0
    Q.append(i)
while Q:
    node=Q.popleft()
    for i in graph[node]:
        if dist[i]==-1:
            dist[i]=dist[node]+1
            Q.append(i)
print(*[dist[d] for d in dests])
```
