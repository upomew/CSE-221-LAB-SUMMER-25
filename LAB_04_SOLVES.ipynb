```python
#TASK_A
inp=input().split()
N,M=int(inp[0]),int(inp[1])
adj_mat=[[0]* N for i in range(N)]
for i in range(M):
  inp1=input().split()
  adj_mat[int(inp1[0]) - 1][int(inp1[1]) - 1]= int(inp1[2])
for r in range(N):
  for c in range(N):
    print(adj_mat[r][c],end=' ')
  print()
```

```python
#Task_B
inp=input().split()
N,M=int(inp[0]),int(inp[1])
ui=list(map(int,input().split()))
vi=list(map(int,input().split()))
wi=list(map(int, input().split()))
adj_lst={j:[] for j in range(1,N+1)}
for i in range(M):
  adj_lst[ui[i]].append((vi[i],wi[i]))
for j in range(1,N+1):
    print(f'{j}:',end='')
    flag=True
    for u,w in adj_lst[j]:
        print(f' ({u},{w})',end='')
        flag=False
    print()
```

```python
#task_D
inp=input().split()
N,M=int(inp[0]),int(inp[1])
graph=[]
for i in range(N):
   graph.append([])
ui=list(map(int,input().split()))
vi=list(map(int,input().split()))
for j in range(M):
  u,v=ui[j]-1,vi[j]-1
  graph[u].append(v)
  graph[v].append(u)
c=0
for i in range(N):
    if len(graph[i])%2!=0:
        c+=1
if c==0 or c==2:
  print('YES')
else:
  print("NO")
```

```python
#Task_E
inp= input().split()
N,M= int(inp[0]),int(inp[1])
u = list(map(int, input().split()))
v = list(map(int, input().split()))
indeg, outdeg = [0] * N, [0] * N
res = []
for i in range(M):
  outdeg[u[i]-1]+= 1
  indeg[v[i]-1]+= 1
for i in range(N):
    diff= indeg[i] - outdeg[i]
    res.append(diff)
for i in res:
  print(i,end=' ')
```

```python
#Task_F
N = int(input())
inp=input().split()
x,y=int(inp[0]),int(inp[1])
possible_moves= []
for i in [-1,0,1]:
    for j in [-1,0,1]:
        if not (i==0 and j==0):
            possible_moves.append((i,j))
V_moves= []
for X,Y in possible_moves:
  m1,m2= x+X,y+Y
  if 1<=m1<=N and 1<=m2<=N:
    V_moves.append((m1,m2))
print(len(V_moves))
for i, j in V_moves:
  print(i,j)
```

```python
#Task_G
def The_Knights_of_Königsberg():
    N,M,K = map(int,input().split())
    pos= []
    check= [[False]*(M+1) for i in range(N+1)]
    for i in range(K):
        x,y = map(int,input().split())
        pos.append((x,y))
        check[x][y]= True
    possible_moves= []
    for i in [-2,-1,1,2]:
     for j in [-2,-1,1,2]:
        if abs(i)!=abs(j):
            possible_moves.append((i,j))
    for i in range(K):
     x1,y1= pos[i]
     for i,j in possible_moves:
        m1,m2 = x1+i,y1+j
        if 1<=m1<=N and 1<=m2<=M:
            if check[m1][m2]:
                        print('YES')
                        return
    print('NO')
The_Knights_of_Königsberg()
```

```python
#Task_H
def gcd(a,b):
    while b:
        a,b= b,a%b
    return a
inp= input().split()
N,Q= int(inp[0]),int(inp[1])
graph =[[] for x in range(N+1)]
queries= []
for i in range(Q):
    a,b = map(int,input().split())
    queries.append((a,b))
for i in range(1,N+1):
  for j in range(i+1,N+1):
    if (i!=j) and (gcd(i,j)==1):
      graph[i].append(j)
      graph[j].append(i)
for X,K in queries:
  if K<=len(graph[X]):
      print(graph[X][K-1])
  else:
      print(-1)
```
