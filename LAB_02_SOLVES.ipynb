```python
#Task_F
inp1= input().split()
N,K= int(inp1[0]),int(inp1[1])
inp2= input().split()
arr= []
for i in range(N):
    arr.append(int(inp2[i]))
dict ={}
max_size,i,j,t = 0,0,0,0
while i< N:
    temp=arr[i]
    if temp in dict:
        if dict[temp]== 0:
            t+= 1
        dict[temp]+= 1
    else:
        dict[temp]= 1
        t+= 1
    while t> K:
        dict[arr[j]]-= 1
        if dict[arr[j]]==0:
            t-= 1
        j+= 1
    if i-j+1> max_size:
        max_size=i-j+1
    i+= 1
print(max_size)
```
