```python
#Task_B
test=int(input())
res=[]
for i in range (test):
    n=input()
    arr=n.split()
    if arr[2]=='+':
        temp=(float(arr[1]) + float(arr[3]))
        res.append(temp)
    elif arr[2]=='-':
        temp=(float(arr[1]) - float(arr[3]))
        res.append(temp)
    elif arr[2]=='*':
        temp=(float(arr[1]) * float(arr[3]))
        res.append(temp)
    elif arr[2]=='/':
        temp=(float(arr[1]) / float(arr[3]))
        res.append(temp)
for j in (res):
    print('{:.6f}'.format(j))
```

```python
#Task_E
def bubble(arr):
    for i in range (len(arr)):
        for j in range (len(arr)-1):
            if arr[j]>arr[j+1]:
              arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
size=int(input())
n=input()
temp=n.split()
arr=[]
res = []
for i in temp:
    arr.append(int(i))
even_arr=[]
odd_arr=[]
for i in range(size):
    if i%2==0:
        even_arr.append(arr[i])
    else:
        odd_arr.append(arr[i])
E=bubble(even_arr)
O=bubble(odd_arr)
x,y =0,0
for i in range(size):
    if i%2==0:
        res.append(E[x])
        x+=1
    else:
        res.append(O[y])
        y+=1
if all(res[i]<=res[i+1]for i in range(size-1)):
    print("YES")
else:
    print("NO")
```

```python
#Task_G
N = int(input())
ID = input()
ID_arr=ID.split()
M = input()
M_arr=M.split()
st=[]
for i in range(N):
    st.append((int(ID_arr[i]), int(M_arr[i])))
count= 0
for j in range(len(st)):
        max_mark = j
        for k in range(j + 1, len(st)):
            if st[k][1] > st[max_mark][1]:
                max_mark = k
            elif st[k][1] == st[max_mark][1] and st[max_mark][0] > st[k][0]:
                max_mark = k
        if max_mark != j:
            st[j], st[max_mark] = st[max_mark], st[j]
            count += 1
print('Minimum swaps:',count)
for i in st:
    print(f'ID: {i[0]} Mark: {i[1]}')
```
