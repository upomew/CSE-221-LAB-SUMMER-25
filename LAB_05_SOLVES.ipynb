```python
#TASK_A
from collections import deque
N,M= map(int,input().split())

adj = [[] for i in range (N+1)]

for i in range (M):
    u,v = map(int,input().split())
    adj[u].append(v)
    adj[v].append(u)

for i in adj:
    i.sort()

visited = [False]*(N+1)
queue= deque()
res=[]

queue.append(1)
visited[1]=True

while queue:
    node=queue.popleft()
    res.append(node)
    for connected in adj[node]:
        if not visited[connected]:
            visited[connected]=True
            queue.append(connected)
for x in res:
  print(x, end=' ')
```

```python
#Task_B
from collections import defaultdict
N,M= map(int,input().split())
u_list=list(map(int,input().split()))
v_list=list(map(int,input().split()))
G= defaultdict(list)

for i in range (M):
    u = u_list[i]
    v= v_list[i]
    G[u].append(v)
    G[v].append(u)

for i in G:
    G[i].sort()
    G[i].reverse()

visited = [False]*(N+1)
Stack=[1]
res=[]

while Stack:
    node=Stack.pop()
    if not visited[node]:
            visited[node]=True
            res.append(node)
            for connected in (G[node][::-1]):
                 if not visited[connected]:
                    Stack.append(connected)
for x in res:
  print(x, end=' ')
```

```python
#task_D
from collections import deque
N,M,S,D,K =map(int,input().split())
adj = [[] for i in range (N+1)]

for i in range (M):
    u,v = map(int,input().split())
    adj[u].append(v)

for i in adj:
   i.sort()

def bfs(St,Dt):
    visited = [-1]*(N+1)
    p_queue=[-1]*(N+1)
    queue=deque()

    queue.append(St)
    visited[St]=0

    while queue:
      node=queue.popleft()
      for connected in adj[node]:
         if visited[connected]== -1:
            visited[connected]=visited[node]+1
            p_queue[connected]=node
            queue.append(connected)
    if visited[Dt]==-1:
     return -1
    else:
     res=[]
     temp=Dt
     while temp!=-1:
        res.append(temp)
        temp= p_queue[temp]
     res.reverse()
     return res
des1= bfs(S,K)
des2= bfs(K,D)
if des1==-1 or des2==-1:
   print(-1)
else:
    result = des1+des2[1:]
    print(len(result)-1)
    for i in result:
     print(i,end=' ')
```

```python
#Task_E
N,R= map(int,input().split())
adj= [[] for i in range (N+1)]
for i in range (N-1):
    u,v=map(int,input().split())
    adj[u].append(v)
    adj[v].append(u)
for i in adj:
    i.sort(reverse=True)
visited=[False]*(N+1)
sub_size=[0]*(N+1)
stack = [(R,-1,0)]
while stack:
    node,p,s= stack.pop()
    if s==0:
        stack.append((node,p,1))
        for connected in adj[node]:
            if connected!=p:
                stack.append((connected,node,0))
    else:
        sub_size[node]=1
        for connected in adj[node]:
            if connected!=p:
                sub_size[node]+= sub_size[connected]
Q = int(input())
for i in range(Q):
    X = int(input())
    print(sub_size[X])
```

```python
#Task_F
N,M=map(int,input().split())
adj= [[]for i in range (N+1)]
for i in range (M):
        u,v=map(int,input().split())
        adj[u].append(v)
visited= [0]*(N+1)
def cycle(start):
    stack=[(start, 0)]
    while stack:
        u,idx= stack[-1]
        if visited[u]== 0:
            visited[u] = 1
        if idx<len(adj[u]):
            v= adj[u][idx]
            stack[-1]=(u,idx+1)
            if visited[v]== 0:
                stack.append((v,0))
            elif visited[v]==1:
                return True
        else:
            visited[u]= 2
            stack.pop()
    return False

for i in range(1,N+1):
    if visited[i]== 0:
        if cycle(i):
            print("YES")
            exit()
print("NO")
```

```python
#Task_G
from collections import deque
R,H= map(int,input().split())
G= []
for i in range(R):
    G.append(list(input().strip()))

Dias_count= 0

def BFS(x,y,R,H):
    Q= deque()
    Q.append((x,y))
    Dias=0
    if G[x][y]=="D":
        Dias+=1
    G[x][y]= "#"
    position=[(-1,0),(1,0),(0,-1),(0,1)]

    while Q:
        x1,y1= Q.popleft()
        for p1,p2 in position:
            move1,move2 = x1+p1, y1+p2
            if 0<=move1<R and 0<=move2<H:
                if G[move1][move2]!="#":
                    if G[move1][move2]=="D":
                        Dias+= 1
                    G[move1][move2]="#"
                    Q.append((move1,move2))
    return Dias

for x in range(R):
    for y in range(H):
        if G[x][y]!= "#":
            Dias_count= max(Dias_count,BFS(x,y,R,H))

print(Dias_count)
```
