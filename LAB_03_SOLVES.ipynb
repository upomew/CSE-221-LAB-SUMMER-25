```python
#Task_A
N=int(input())
arr= list(map(int,input().split()))
inv=0
def merge(a,b):
  global inv
  i,j=0,0
  arr=[]
  while i <len(a) and j<len(b):
    if a[i]>b[j]:
        arr.append(b[j])
        inv+= len(a)-i
        j+=1
    else:
        arr.append(a[i])
        i+=1
  if i<len(a):
     for k in range(i,len(a)):
        arr.append(a[k])
  elif j<len(b):
    for k in range(j,len(b)):
        arr.append(b[k])
  return arr

def mergeSort(arr):
    if len(arr) <= 1:
        return arr
    else:
        mid= len(arr)//2
        a1= mergeSort(arr[:mid:])
        a2= mergeSort(arr[mid::])
        return merge(a1,a2)
res= mergeSort(arr)
print(inv)
for i in res:
   print(i,end=' ')
```

```python
#Task_B
import bisect
def merge_inv(arr,l,r):
    if l>=r:
        return 0
    m =(l+r)// 2
    inv =merge_inv(arr,l,m) + merge_inv(arr,m+1,r)
    a=sorted(arr[l:m+ 1])
    b=arr[m+1:r+1]
    for j in b:
        i =bisect.bisect_right(a, j*j)
        inv +=len(a)-i
    arr[l:r+1] =sorted(arr[l:r+1])
    return inv
N =int(input())
arr = list(map(int,input().split()))
print(merge_inv(arr,0,N-1))
```

```python
#Task_D
mod=10**9+7
def mat_mult(arr1,arr2):
    res=[[0,0],[0,0]]
    for i in range(2):
        for j in range(2):
            val=0
            for k in range(2):
                val+=arr1[i][k]*arr2[k][j]
            res[i][j]=val%mod
    return res
def fast_mat(arr,x):
    result=[[1,0],[0,1]]
    while x>0:
        if x%2==1:
            result=mat_mult(result,arr)
        arr=mat_mult(arr,arr)
        x=x//2
    return result
T=int(input())
for i in range (T):
    a11,a12,a21,a22=map(int,input().split())
    arr=[[a11,a12],[a21,a22]]
    x=int(input())
    r=fast_mat(arr,x)
    print(r[0][0],r[0][1])
    print(r[1][0],r[1][1])

```

```python
#Task_E
T = int(input())
inp= [list(map(int, input().split()))for x in range(T)]
def fast_series(a,n,m):
  tempN,tempA=n,a%m
  c=1
  a=tempA
  res=a
  while n>1:
      c+=c
      res=(res+(res*a))% m
      a=(a**2)%m
      n//=2
  if tempN >c:
      res =(res+a*(fast_series(tempA,tempN-c, m)))%m
  return res
res= []
for a,n,m in inp:
  res.append(fast_series(a,n,m))
for i in res:
   print(i)
```

```python
#Task_F
N= int(input())
arr= list(map(int,input().split()))
def orderingBT(arr):
  a=[]
  def helper(l,h):
    if l>h:
        return
    m=(l+h)//2
    a.append(arr[m])
    helper(l, m-1)
    helper(m+1, h)
  helper(0,len(arr)-1)
  return a
res=orderingBT(arr)
for i in res:
   print(i,end=' ')
```

```python
#Task_G
N =int(input())
a =list(map(int,input().split()))
b =list(map(int, input().split()))
def tree(a,b):
    idx_dct={}
    for i in range(N):
        idx_dct[a[i]]= i
    pO_idx=[0]
    def build_pO(start,end):
        if start> end:
            return []
        r=b[pO_idx[0]]
        pO_idx[0]+= 1
        root= idx_dct[r]
        l_subT,r_subT= build_pO(start, root-1),build_pO(root+1, end)
        return l_subT+r_subT+[r]
    return build_pO(0,N-1)
res=tree(a,b)
for i in res:
    print(i,end=' ')
```

```python
#Task_H
N =int(input())
a =list(map(int,input().split()))
b =list(map(int, input().split()))
def tree_r(a,b):
    idx_dct={}
    for i in range(N):
        idx_dct[a[i]]= i
    pr_idx=[N-1]
    def build_pr(start,end):
        if start> end:
            return []
        r=b[pr_idx[0]]
        pr_idx[0]-= 1
        root= idx_dct[r]
        r_subT,l_subT=build_pr(root+1, end), build_pr(start, root-1)
        return [r]+l_subT+r_subT
    return build_pr(0,N-1)
res=tree_r(a,b)
for i in res:
    print(i,end=' ')
```
