```python
#Task_B
import heapq
N,M,S,T= map(int,input().split())
visited={i:0 for i in range (N+1)}
edges = {i:[] for i in range(1,N+1)}
for x in range(M):
  u,v,w=list(map(int,input().split()))
  edges[u].append((v,w))
def dijkstra(edges,S):
  dist_src = {i:float('inf') for i in range(1,N+1)}
  visited = {i:0 for i in range(1,N+1)}
  parent = {i:-1 for i in range(1,N+1)}
  pq=[(0,S)]
  dist_src[S]= 0
  while pq:
    d,ver= heapq.heappop(pq)
    visited[ver] = 1
    for node, W in edges[ver]:
      if dist_src[ver]+W<dist_src[node]:
        dist_src[node]=dist_src[ver]+W
        parent[node]=ver
        heapq.heappush(pq,(dist_src[node],node))
  return dist_src
if S==T:
  print(0,S)
else:
  a_d= dijkstra(edges,S)
  b_d= dijkstra(edges,T)
  inf=float('inf')
  Min=inf
  dist=inf
  for i in range(1,N+1):
    if a_d[i]!=inf and b_d[i]!=inf:
      res_t=max(a_d[i],b_d[i])
      if res_t<Min:
        Min=res_t
        dist=i
      elif res_t==Min and i<dist:
        dist=i
  if dist==inf:
    print(-1)
  else:
    print(Min,dist)
```

```python
#task_D
import heapq
N,M,S,D= map(int,input().split())
w=list(map(int,input().split()))
edges = {i:[] for i in range(1,N+1)}
for x in range(M):
  u,v=list(map(int,input().split()))
  edges[u].append(v)
def dijkstra(edges,S,w,D):
  dist_src = {i:float('inf') for i in range(1,N+1)}
  pq=[(w[S-1],S)]
  dist_src[S]= w[S-1]
  while pq:
    W,ver= heapq.heappop(pq)
    if dist_src[ver]<W:
        continue
    if ver== D:
        return W
    for node in edges[ver]:
        if dist_src[ver]+w[node-1]<dist_src[node]:
            dist_src[node]=dist_src[ver]+w[node-1]
            heapq.heappush(pq,(dist_src[node],node))
  return -1
print(dijkstra(edges,S,w,D))
```

```python
#Task_E
import heapq
N,M= map(int,input().split())
u=list(map(int,input().split()))
v= list(map(int,input().split()))
w=list(map(int,input().split()))
edges = {i:[] for i in range(1,N+1)}
for x in range(M):
  edges[u[x]].append((v[x], w[x], w[x]%2))
def dijkstra(edges):
  dist_src = {i:[float('inf'),float('inf')] for i in range(1,N+1)}
  pq=[]
  for n,w,p in edges[1]:
    dist_src[n][p]=w
    heapq.heappush(pq,(w,n,p))
  while pq:
    wei,ver,par=heapq.heappop(pq)
    for node,W,curr in edges[ver]:
       if curr==par:
        continue
       if wei+W<dist_src[node][curr]:
        dist_src[node][curr]=wei+W
        heapq.heappush(pq,(dist_src[node][curr],node,curr))
  res= min(dist_src[N][0],dist_src[N][1])
  if res!= float('inf'):
    return res
  else:
    return -1
print(dijkstra(edges))
```
