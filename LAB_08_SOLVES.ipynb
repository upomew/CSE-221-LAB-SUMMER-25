```python
#TASK_A
def disj(N,K):
    parent=[i for i in range(N+1)]
    size=[1]*(N+1)
    def find(X):
        if parent[X]!=X:
            parent[X]=find(parent[X])
        return parent[X]
    def union(x,y):
        x1=find(x)
        y1=find(y)
        if x1==y1:
            return size[x1]
        if size[x1]<size[y1]:
            x1,y1=y1,x1
        parent[y1]=x1
        size[x1]+=size[y1]
        return size[x1]
    res=[]
    for i in range(K):
        n1,n2= map(int,input().split())
        res.append(union(n1,n2))
    return res
N,K= map(int,input().split())
for x in disj(N,K):
    print(x)
```

```python
#Task_B
N,M= map(int,input().split())
edges= [tuple(map(int,input().split())) for i in range(M)]
def mst(N,edges):
    edges=sorted(edges,key=lambda x:x[2])
    parent=[i for i in range(N+1)]
    size=[1]*(N+1)
    def find(X):
        if parent[X]!=X:
            parent[X]=find(parent[X])
        return parent[X]
    def union(x,y):
        x1=find(x)
        y1=find(y)
        if x1==y1:
            return False
        if size[x1]<size[y1]:
            x1,y1=y1,x1
        parent[y1]= x1
        size[x1]+=size[y1]
        return True
    c,edg_cnt= 0,0
    for u,v,w in edges:
        if union(u,v):
            c+=w
            edg_cnt+=1
            if edg_cnt==N-1:
                break
    return c
print(mst(N,edges))
```

```python
#Task_C
N,M= map(int,input().split())
edges=[]
for x in range(M):
    u,v,w = map(int,input().split())
    edges.append((w,u,v))
def secondmst(N,M,edges):
  edges.sort()
  def mst(exE=None):
    parent=list(range(N + 1))
    def find(X):
      while parent[X]!=X:
          parent[X]=parent[parent[X]]
          X=parent[X]
      return X
    c=0
    edg_cnt=[]
    for i,(w,u,v) in enumerate(edges):
      if exE!= None and i==exE:
        continue
      u1=find(u)
      v1=find(v)
      if u1!= v1:
        parent[v1]= u1
        c+= w
        edg_cnt.append((w,u,v))
    if len(edg_cnt)!= N-1:
      return None
    return c
  best_mst=mst()
  if best_mst is None:
    return -1
  second_best=float('inf')
  n_parent=list(range(N + 1))
  def find_best(X):
    while n_parent[X]!= X:
      n_parent[X]= n_parent[n_parent[X]]
      X= n_parent[X]
    return X
  mst_edge=[]
  for i,(w,u,v) in enumerate(edges):
    u2=find_best(u)
    v2=find_best(v)
    if u2!=v2:
      n_parent[v2]=u2
      mst_edge.append(i)
  for excl in mst_edge:
    curr= mst(excl)
    if curr!= None and best_mst<curr< second_best:
      second_best= curr
  non_mst= [x for x in range(M) if x not in mst_edge]
  for x in non_mst:
    w,u,v=edges[x]
    temp_parent=list(range(N + 1))
    def find_temp(X):
      while temp_parent[X]!=X:
        temp_parent[X]=temp_parent[temp_parent[X]]
        X=temp_parent[X]
      return X
    cost= w
    temp_parent[u]= v
    for i in mst_edge:
      wi,ui,vi= edges[i]
      u3= find_temp(ui)
      v3= find_temp(vi)
      if u3!=v3:
        temp_parent[v3]= u3
        cost+=wi
    if best_mst<cost<second_best:
      second_best=cost
  return -1 if second_best==float('inf') else second_best
print(secondmst(N,M,edges))
```
